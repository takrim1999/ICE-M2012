## Problem:
Y = mX, Y = 9, X=3, what is the m?

```python
# naive approach
error = 0.S3
Y = 9
X = 3
# let, m = 0, we will increase it
m = 0
while True:
    y = m*X
    e = ((Y-y)/Y)*100
    print(f"X is {X}, m is {m} so, Y is {Y}, error: {e}%")
    m = (m+X)/2
    print(f"New m: {m}")
    if round(e) <= error:
        print("Finally converged!")
        print(f"Selected {m} with error {e}%")
        break
```
